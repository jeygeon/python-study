
# 1. Python이란?

Python은 1991년 처음 공개된 고급 프로그래밍 언어이다.

간결한 문법과 다양한 라이브러리를 지원하며,
웹 개발, 데이터 분석, 인공지능, 자동화 등
다양한 분야에서 활용된다.

## Python의 주요 특징

### 1. 인터프리터 언어
- 인터프리터를 통해 코드를 실행한다.
- CPython은 소스 코드를 바이트코드로 컴파일한 후 실행한다.

### 2. 동적 타이핑
- 변수를 선언할 때 자료형을 명시하지 않아도 된다.
- 변수에 다른 자료형의 객체를 다시 할당할 수 있다.

### 3. 간결한 문법
- 들여쓰기를 이용해 코드 블록을 구분한다.
- 중괄호와 세미콜론을 일반적으로 사용하지 않는다.

### 4. 풍부한 라이브러리
- NumPy: 수치 계산
- Pandas: 데이터 분석
- Matplotlib: 데이터 시각화
- Scikit-learn: 머신러닝
- Requests: HTTP 통신

## Python의 주요 활용 분야

- 웹 백엔드 개발
- 데이터 분석
- 인공지능 및 머신러닝
- 업무 자동화
- 웹 크롤링


# 2. Python 실행 환경

Python 코드를 실행하는 대표적인 방법은 다음과 같다.

| 실행 방식 | 설명 | 파일 형식 |
|---|---|---|
| 인터프리터 | 명령어를 직접 입력하고 실행 | 없음 |
| 스크립트 | Python 파일을 실행 | .py |
| Jupyter Notebook | 셀 단위로 코드 실행 | .ipynb |

## Jupyter Notebook 구성

- Notebook: 코드와 설명을 작성하는 문서
- Kernel: 코드를 실제로 실행하는 프로세스
- Interpreter: Python 코드를 처리하고 실행하는 프로그램

### Kernel 특징
- 실행한 변수와 상태를 메모리에 유지한다.
- 여러 셀에서 동일한 변수를 사용할 수 있다.
- Kernel을 재시작하면 메모리의 변수들이 초기화된다.


````python
import sys

# Python 버전 확인
print("Python Version:", sys.version)

# Python 실행 파일 경로 확인
print("Python Path:", sys.executable)
````

````python
# 기본적인 출력 확인
print("Hello, Python!")
print("Python 공부를 시작합니다.")
print(10 + 20)
````
