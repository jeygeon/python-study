### 1. 변수(Variable)
- 데이터를 참조하기 위해 사용하는 이름
- Python 변수는 객체를 참조한다.
- 변수에 객체를 재할당할 수 있다.

```python
# 변수 name은 "Python"이라는 객체를 참고하고 있다.
name = "Python"
```

```python
# a는 10을 참고하고 있고, b는 a를 참조하고 있다.
# 결국 둘은 같은 객체를 참고하고 있는 상태이다.
a = 10
b = a

print(a)
print(b)

# 실제 둘의 객체 고유 식별자를 조회하면 같은 주소를 참조하는 것을 알 수 있다.
# id()를 통해 해당 객체의 식별 값 확인이 가능하다.
print(id(a))
print(id(b))
```

### 2. Mutable
객체 내부의 값을 변경할 수 있다.

- list
- dict
- set

### 3. Immutable
객체 생성 후 내부 값을 변경할 수 없다.

- int
- float
- str
- tuple
- bool

```python
a = 20
b = a

print(a)
print(b)

# 같은 객체 참조
print(id(a))
print(id(b))

a = 10
print(a)
print(b)

# 서로 다른 객체 참조
print(id(a))
print(id(b))
```

```python
a = [1, 2, 3]

print(id(a))

a.append(4)

print(a)
print(id(a))
```

```python
# Immutable
a = 10
before = id(a)

a += 1

after = id(a)

print("Immutable")
print(before)
print(after)
print(before == after)


# Mutable
b = [1, 2, 3]
before = id(b)

b.append(4)

after = id(b)

print("Mutable")
print(before)
print(after)
print(before == after)
```

### 4. 객체 비교

| 연산자 | 의미 |
|---|---|
| == | 값 비교 |
| is | 객체 동일성 비교 |

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b) # 단순 값 비교 같음 True
print(a is b) # 객체 동일 하지 않음 False
```
