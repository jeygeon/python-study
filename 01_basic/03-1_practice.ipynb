```python
# 기본적인 데이터 연산 배운 기념 간단한 문제 만들기 (함수 사용은 구글링으로 사용법만 확인)

# 1. 사용자의 아이디 비밀번호를 입력받아서
# 2. 아이디, 비밀번호가 정상적으로 입력됐는지 확인
# 3. 해당 사용자가 일반 사용자인지 관리자 사용자인지 확인

user_id = input("아이디를 입력하세요 : ")
user_password = input("비밀번호를 입력하세요 : ")

# 입력받은 값이 관리자 계정인지 확인 (간단하게 id가 admin인지만..)
admin_user_id = "admin"

def isValidUser():
    return bool(user_id.strip() and user_password.strip())

def isAdminUser():
    return user_id == admin_user_id

if not isValidUser():
    print("입력값이 올바르지 않습니다.")
elif isAdminUser():
    print("관리자입니다.")
else: 
    print("일반사용자입니다.")
```
