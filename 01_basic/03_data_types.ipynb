# Python 기본 자료형 요약

일단 자바와는 달리 파이썬에서는 변수 선언을 할 떄 명시적으로 타입을 지정할 필요없이 동적으로 타입을 정해준다

## 1. 기본 자료형

| 자료형 | 설명 | 예시 |
|---|---|---|
| int | 정수 | 10 |
| float | 실수 | 3.14 |
| str | 문자열 | "Python" |
| bool | 참/거짓 | True, False |
| NoneType | 값이 없음 | None |

## 2. 자료형 확인

- `type(value)`: 객체의 실제 자료형 확인
- `isinstance(value, type)`: 지정한 클래스의 인스턴스인지 확인

```python
a = 10
b = 3.14
c = "Python"
d = True
e = None

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
```

```python
number = 10

print(isinstance(number, int))
print(isinstance(number, str))
print(isinstance(number, float))
```

`type()`과 `isinstance()`는 모두 객체의 타입이 뭔지 확인할 수 있지만 두 개의 차이점은 특정 객체를 상속받냐 안받냐에 따라 사용이 달라질 수 있다.

```python
class Animal:
    pass

class Dog(Animal):
    pass

dog = Dog()

print(type(dog) == Animal)
print(isinstance(dog, Animal))
```

> 여기서 `type(dog) == Animal`은 실제 `dog`가 `Animal`의 타입인지 확인을 하고 <br/>
`isinstance(dog, Animal)`은 `dog`가 `Animal` 타입인지 또는 그것을 상속받고 있는 타입인지 확인 하는 차이가 있다.

## 3. 숫자 자료형

- `int`: 정수
- `float`: 실수
- `/`: 나눗셈
- `//`: 몫
- `%`: 나머지
- `**`: 거듭제곱

`float` 타입의 소수점 계산에서는 일부 계산이 의도한 것과 다르게 결과가 나올 수 있다.

```python
a = 0.1
b = 0.2

print(a + b)
print(a + b == 0.3)
```

이런 현상이 발생되는 이유는 컴퓨터는 모든 10진수의 소수를 2진수로 변환할 수 없다. <br/>
에를 들어 0.1을 2진수로 바꾸게 되면 무한이 반복되는 소수가 되는데, 이를 근사값으로 저장되고 계산과정에서 미세한 오차가 발생 할 수 있다.<br/>
이를 정확하게 방지하려면 `decimal.Decimal`을 사용하면 된다.

```python
from decimal import Decimal

# 문자열을 이용해 Decimal을 생성하면 10진수 값을 의도한 그대로 표현 가능하다.
a = Decimal("0.1")
b = Decimal("0.2")

print(a + b)
print(type(a + b))
```



## 4. 문자열

- `str`은 Immutable 자료형
- 인덱싱: `text[0]`
- 음수 인덱싱: `text[-1]`
- 슬라이싱: `text[0:5]`
- f-string: `f"Hello {name}"`

주요 메서드:
- `strip()`
- `lower()`
- `upper()`
- `replace()`
- `find()`
- `split()`
- `join()`


```python
# str은 문자를 표현하는 데이터 타입이다.
name = "Python"
language = 'Java'

print(name)
print(language)
print(type(name))

# 큰 따옴표와 작은 따옴표 둘다 사용할 수 있다.
a = "Hello"
b = 'Hello'

print(a == b)
```

```python
# 문자열은 Immutable 자료형이기 때문에 수정이 불가능하다. 그래서 수정하고 싶다면 새로운 객체를 생성해야 한다.

text = "Python"
text = "J" + text[1:]

print(text)
```

```python
# 문자열 인덱싱 (Indexing)
# 문자열의 인덱스를 통해 해당 문자열이 뭔지 알 수 있다.
# 앞에서부터 시작은 0부터
# 뒤에서부터 시작은 -1부터
text = "Python"

print(text[0])
print(text[1])
print(text[5])
print(text[-1])
```

```python
# 문자열 슬라이싱 (Slicing)
# 문자열의 일부분을 잘라내는 방식이다.
text = "Python Programming"

print(text[0:6])
print(text[7:18])
print(text[:6])
print(text[7:])
print(text[::-1])

# 기본 문법은 아래와 같다.
# 문자열[시작인덱스:종료인덱스:증가폭]

# 마지막 인덱스는 포함하지 않는다.
# 예를 들어 text[0:6]은 0번부터 5번까지 가져온다.
```

```python
# 자주 사용하는 문자열 메소드
text = "  Hello Python  "

print(text.strip()) # 양쪽 공백 제거
print(text.lower()) # 전체 문자열 소문자로 변경
print(text.upper()) # 전체 문자열 대문자로 변경
print(text.replace("Python", "Java")) # 특정 문자열 변경
print(text.find("Python")) # 문자열 위치 찾기
print(text.count("o")) # 문자열 개수
```

```python
# 문자열 포멧팅
name = "Kim"
age = 30

print("이름: {}, 나이: {}".format(name, age))
print(f"이름: {name}, 나이: {age}")
```
