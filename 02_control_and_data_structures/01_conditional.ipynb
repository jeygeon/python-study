#### 1. Conditional (조건문)
조건문이란 특정 조건이 참(True)인지 거짓(False)인지 판별해서 서로 다른 코드를 실행하는 문법이다.

#### 2. if문
가장 기본적인 조건문이다
```python
if 조건식:
    실행할 코드
```

조건식이 True면 아래 코드가 실행되고 아니면 실행되지 않는다.

````python
# 나이 확인하기
age = 25
if age >= 19:
    print("성인입니다")
````

#### 2. if ~ else문
조건이 참일때와 거짓일때 각각 다른 코드를 실행하고 싶을 때 사용한다.
```python
if 조건식:
    조건식이 True일 때 실행
else:
    조건식이 False일 때 실행
```

````python
# 합격 여부 확인하기
score = 90
if score >= 80:
    print("합격입니다.")
else:
    print("불합격입니다.")
````

#### 2. if ~ elif ~ else문
조건이 여러개일 때 사용하는 구조이다

```python
if 조건식1:
    조건식1이 True일 때 실행
elif:
    조건식2가 True일 때 실행
elif:
    조건식3이 True일 떄 실행
else:
    모두 False일 때 실행
```

````python
# 성적에 따른 학점 계산
score = 85
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
````

````python
# 조건식에서는 and, or, not같은 연산자 사용도 가능하다.
user_id = "admin"
user_password = "1234"

if user_id == "admin" and user_password == "1234":
    print("로그인 성공")
else:
    print("로그인 실패")
````
