#### 1. Loop(반복문)
반복문은 동일하거나 유사한 작업을 여러 번 실행할 때 사용하는 문법이다.
- 1부터 100까지 숫자 출력
- 학생 30명의 평균 점수 계산
- 회원 목록을 하나씩 조회

반복문은 `for`, `while` 두 가지가 있다.

#### 2. for문

```python
for 변수 in 반복데이터:
    실행할 코드
```

````python
students = ["Kim", "Park", "Lee"]
for student in students:
    print(student)
````

#### 3. range()
`range()` 은 숫자의 범위를 생성하는 함수이다. 반복문과 많이 사용된다.

구조 : `range(start, stop, step)`<br/>
- `start` : 시작 숫자 (기본값 : 0)
- `stop` : 종료 숫자 (기본값 : X)(필수로 있어야 하는 값)(해당값은 미포함 된다)
- `step` : 증가 또는 감소하는 간격 (기본값 : 1)


````python
# 0부터 5까지 출력하기
for i in range(5):
    print(i)
````

````python
# 시작 숫자 지정
for i in range(1, 6):
    print(i)
````

````python
# 증가 간격 조정
for i in range(1, 11, 2):
    print(i)
````

#### 4. while
`while`은 조건식이 참일동안 계속 반복되는 조건문이다.

```python
while 조건식:
    실행할 코드
```

````python
num = 1
while num <= 5:
    print(num)
    num += 1
````

#### 4. break
`break`를 사용하면 반복문을 즉시 종료할 수 있다.

````python
i = 0
while i<=5:
    print(i)
    i += 1
    if i == 3: # 여기서 만약 i가 3이라면 반복문 종료
        break
````

#### 5. continue
`continue`를 사용하면 반복문 도중 현재 구간만 스킵하고 다음으로 넘어갈 수 있다.

````python
nums = [1, 2, 3, 4, 5]
for num in nums:
    if num == 3: # 여기서 num이 3이면 출력하지않고 다음 반복으로 스킵한다.
        continue
    print(num)
````

````python
# 반복문안에 반복문을 사용하여 중첩으로도 사용이 가능하다.
# 구구단 1단 출력해보기
for i in range(1, 2):
    for j in range(1, 10):
        print(f"{i} x {j} = {i * j}")
````

#### 6. enumerate()
`enumerate`를 사용하면 리스트를 반복하면서 인덱스를 같이 표기 할 수 있다.

````python
members = ["Kim", "Lee", "Park"]
for index, member in enumerate(members):
    print(index, member)
````

````python
# 시작인덱스 변경도 가능하다.
members = ["Kim", "Lee", "Park"]
for index, member in enumerate(members, start = 1):
    print(index, member)
````

#### 7. zip()
`enumerate`를 사용하면 여러 리스트를 반복문에 같이 사용할 수 있다.

````python
students = ["Kim", "Lee", "Park"]
scores = [80, 30, 70]

for student, score in zip(students, scores):
    print(f"{student} : {score}점")
````
