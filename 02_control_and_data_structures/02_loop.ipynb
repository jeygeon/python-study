#### 1. Loop(반복문)
반복문은 동일하거나 유사한 작업을 여러 번 실행할 때 사용하는 문법이다.
- 1부터 100까지 숫자 출력
- 학생 30명의 평균 점수 계산
- 회원 목록을 하나씩 조회

반복문은 `for`, `while` 두 가지가 있다.

#### 2. for문

```python
for 변수 in 반복데이터:
    실행할 코드
```

````python
students = ["Kim", "Park", "Lee"]
for student in students:
    print(student)
````

#### 3. range()
`range()` 은 숫자의 범위를 생성하는 함수이다. 반복문과 많이 사용된다.

구조 : `range(start, stop, step)`
