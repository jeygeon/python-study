#### 1. List
`list`는 여러개의 데이터를 한번에 담을 수 있는 자료형이다.
- 데이터의 순서가 있다.
- 중복 데이터 허용한다.
- 각각의 데이터에 인덱스가 있다.

```python
# 숫자 리스트
numbers = [10, 20, 30, 40, 50]

# 문자열 리스트
fruits = ["apple", "banana", "orange"]

# 다양한 자료형 혼합
mixed = [10, "hello", 3.14, True]

# 빈 리스트
empty_list = []

print(numbers)
print(fruits)
print(mixed)
print(empty_list)
```

```python
# 리스트의 데이터에는 각각의 인덱스가 존재하며 각 인덱스를 통해 데이터에 접근도 가능하다.
# 처음은 0부터 시작하고, 마지막부터 접근 시 -1부터 시작
fruits = ["apple", "banana", "orange", "grape"]

# 양수 인덱스
print(fruits[0])  # apple
print(fruits[1])  # banana
print(fruits[2])  # orange

# 음수 인덱스
print(fruits[-1])  # grape
print(fruits[-2])  # orange
```

데이터 슬라이싱 기본 문법
`list[start:end:step]`

- start: 시작 인덱스 (포함)
- end: 종료 인덱스 (미포함)
- step: 데이터 추출 간격

```python
# 슬라이싱을 통해 원하는 범위의 데이터 추출도 가능하다.
numbers = [10, 20, 30, 40, 50, 60]

# 기본 슬라이싱
print(numbers[1:4])   # [20, 30, 40]

# 처음부터 특정 위치까지
print(numbers[:3])    # [10, 20, 30]

# 특정 위치부터 끝까지
print(numbers[3:])    # [40, 50, 60]

# 전체 리스트
print(numbers[:])     # [10, 20, 30, 40, 50, 60]

# 일정한 간격으로 추출
print(numbers[::2])   # [10, 30, 50]

# 역순으로 추출
print(numbers[::-1])  # [60, 50, 40, 30, 20, 10]
```

```python
# len()을 통해 리스트의 길이도 확인이 가능하다.
numbers = [10, 20, 30, 40, 50]

print(len(numbers))
```

#### 2. List 데이터 수정
List는 Mutable 자료형이기 때문에 데이터 수정이 가능하다.

```python
students = ["Lee", "Kim", "Park"]
students[1] = "Choi"
print(students)
```

```python
# 리스트 슬라이싱을 통해서도 수정이 가능하다.
students = ["Lee", "Kim", "Park"]
students[1:3] = ["Choi", "Kang"]
print(students)
```

```python
# 특정 데이터 수정
students = ["Lee", "Kim", "Park"]
del students[1]
print(students)

# 데이터  슬라이싱을 통해서도 삭제 가능하다
students = ["Lee", "Kim", "Park"]
del students[1:]
print(students)
```

#### 3. List 메소드
리스트에 자주 사용되는 메소드들이 있다.

```python
# append() : 요소 삽입
numbers = [10, 20, 30, 40]
numbers.append(50)
print(numbers)
```

```python
# extend() : 반복 가능한 객체들을 추가함
numbers = [10]
numbers.extend([20, 30, 40])
print(numbers)
```

```python
# insert() : 지정한 위치에 요소를 추가
numbers = [10, 30, 40]
numbers.insert(1, 20)
print(numbers)
```

```python
# remove() : 특정 요소를 삭제
numbers = [10, 20, 30, 40]
numbers.remove(20)
print(numbers)
```

```python
# pop() : 특정 인덱스의 요소를 삭제하고 반환한다. 값이 없을 경우 제일 마지막을 삭제
numbers = [10, 20, 30, 40]
result1 = numbers.pop(1)
print(f"{numbers}, 반환된 요소 : {result1}")

result2 = numbers.pop()
print(f"{numbers}, 반환된 요소 : {result2}")
```

```python
# clear() : 리스트의 요소를 모두 삭제
numbers = [10, 20, 30]
numbers.clear()
print(numbers)
```

```python
# index() : 특정 값이 처음 존재하는 인덱스를 조회
numbers = [10, 20, 30]
result = numbers.index(10)
print(result)
```
