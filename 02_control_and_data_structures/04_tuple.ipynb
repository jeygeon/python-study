#### 1. tuple
`Tuple(튜플)`은 여러개의 데이터를 한번에 저장할 수 있는 자료형이다.

`List`와 비슷하지만 가장 큰 특징은 `Immutable`이기 때문에 생성한 이후 수정이 안된다는 점이다.
- 데이터의 순서가 있다.
- 중복 데이터 허용한다.
- 각각의 데이터에 인덱스가 있다.
- 생성한 이후 수정이 불가능하다.

```python
# 리스트는 대괄호[]를 사용하지만 튜플은 선연시 괄호()를 사용한다.

# 숫자 Tuple
numbers = (10, 20, 30, 40, 50)

# 문자열 Tuple
fruits = ("apple", "banana", "orange")

# 다양한 자료형 혼합
mixed = (10, "hello", 3.14, True)

# 빈 Tuple
empty_tuple = ()

print(numbers)
print(fruits)
print(mixed)
print(type(numbers))
```

`tuple`은 데이터 수정, 삭제가 안된다는 점 외에는 리스트와 매우 비슷하다

#### 2. Packing과 Unpacking


```python
# 여러개의 데이터를 하나로 묶는 것을 Packing이라고 한다.
person = "Kim", 29, "Developer"
print(person)
```

```python
# Unpacking은 저장된 여러 데이터를 다른 변수에 각각 나누어 할당하는 것을 말한다.
person = ("Kim", 29, "Developer")
name, age, job = person
print(name)
print(age)
print(job)
```

```python
# 리스트와 튜블은 서로 변환이 가능하다.

# Tuple → List
numbers = (10, 20, 30)
number_list = list(numbers)
number_list.append(40)
print(number_list)

# List → Tuple
number_tuple = tuple(number_list)
print(number_tuple)
```
