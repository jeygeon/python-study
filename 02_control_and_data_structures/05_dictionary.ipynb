#### 1. Dictionary(딕셔너리)
딕셔너리는 key와 value를 하나의 쌍으로 저장하는 자료형이다.

리스트나 튜플이 인덱스를 이용해서 value에 접근한다면, 딕셔너리는 key를 이용해서 접근한다.

- key를 이용해서 데이터에 접근한다.
- 추가, 수정, 삭제가 가능하다.
- 하나의 딕셔너리에는 중복된 key를 가질 수 없다.
- value에는 튜플, 리스트, 객체 등 다양한 타입을 가질 수 있다

```python
# 딕셔너리 생성

# 중괄호를 사용
user = {
    "name": "김철수",
    "age": 29
}

# dict() 사용
user = dict(name = "김철수", age = 25)
```

```python
# 딕셔너리
user = {
    "name": "김철수",
    "age": 29,
    "job": "개발자"
}

print(user["name"])
print(user["age"])
```

```python
# 대괄호를 이용해서 value를 가져올 수도 있지만 get()을 사용해서도 가능하다.

user = {
    "name": "김철수"
}

# 존재하지 않는 Key
print(user["age"])         # KeyError 발생
print(user.get("age"))     # None 반환
print(user.get("age", 0))  # 0 반환
```

```python
# 추가, 수정

user = {
    "name": "김철수",
    "age": 29
}

# 데이터 추가
user["job"] = "개발자"

# 데이터 수정
user["age"] = 30

print(user)
```

```python
# 삭제

user = {
    "name": "김철수",
    "age": 29,
    "job": "개발자"
}

# del: 특정 Key 삭제
del user["age"]
print(user)

# pop(): 특정 Key 삭제 후 Value 반환
job = user.pop("job")
print(job)
print(user)

# popitem(): 마지막 항목 삭제 후 반환
item = user.popitem()
print(item)
print(user)

# clear(): 전체 데이터 삭제
user.clear()
print(user)
```

```python
# 딕셔너리 관련 메소드

user = {
    "name": "김철수",
    "age": 29,
    "job": "개발자"
}

# 모든 Key 조회
print(user.keys())

# 모든 Value 조회
print(user.values())

# 모든 Key, Value 쌍 조회
print(user.items())

# Key 존재 여부 확인
print("name" in user)
print("email" in user)

# 데이터 개수 확인
print(len(user))
```

```python
# 딕셔너리 안에 중첩으로 저장도 가능하다.
users = {
    "user1" : {
        "name" : "KIM",
        "age" : 20

    },
    "user2" : {
        "name" : "PARK",
        "age" : "35"
    }
}

# 조회
print(users.get("user1").get("name"))
print(users.get("user2").get("age"))

# 추가
users.get("user1")["location"] = "SEOUL"
print(users.get("user1").get("location"))

# 수정
users.get("user1")["name"] = "LEE"
print(users.get("user1").get("name"))

# 삭제
del users.get("user1")["location"]
print(users.get("user1"))
```
