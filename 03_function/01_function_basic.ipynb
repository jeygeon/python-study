#### 1. Function (함수)
함수란 특정 작업을 수행하는 코드를 하나로 묶어 재사용 할 수 있게 만든 기능이다.

파이썬에서는 일반적으로 함수명을 `snake_case` 작성한다.

자바에서는 함수 내 코드 작성을 `{}`로 감쌌지만 파이썬에서는 `들여쓰기(tab)`으로 구분짓는다

#### 함수 정의
```python
def 함수명():
    실행할 코드
```

#### 정의한 함수 호출
```python
함수명()
```

````python
def hello():
    print("hello world")

hello()
````
