#### 1. Parameter(매개변수)와 Argument(인자)
- `Parameter` : 함수를 정의할 떄 데이터를 저장하는 변수
- `Argument` : 함수를 호출할 때 전달하는 데이터 값

```python
# name은 매개변수
def hello(name):
    print(f"hello {name}")

# 인자로 python을 전달
hello("python")
```

```python
# 인자는 매개변수를 직접 지정해서 전달도 가능하다.
def introduce(name, age, location):
    print(f"이름 : {name}, 나이 : {age}, 사는 곳 : {location}")

introduce(name="KIM", age=25, location="SEOUL")
```

```python
# 매개변수는 디폴트 값을 지정해서 선언도 가능하다.
# 이 경우 함수 호출 시 전달하는 인자가 없을 경우 디폴트 값으로 코드가 실행된다.
def hello(name="python"):
    print(f"hello {name}")

hello()
```

```python
# 가변 위치 인자 (*args) : 함수에 몇개의 인자가 전달 될지 모르는 경우에 사용된다.
# 이 때 인자는 튜플 형태로 전달된다.
def calculate(*args):
    print(type(args))
    print(sum(args))

calculate(10, 20, 30)
calculate(15, 15)
```

```python
# 가변 키워드 인자 (**kwargs) : 인자의 갯수가 정해지지 않은 키워드를 전달할 떄 사용한다.
# 이 때 인자는 딕셔너리 형태로 전달된다.
def introduce(**kwargs):
    print(type(kwargs))
    print(kwargs)

introduce(name = "KIM", age = 25, location = "SEOUL")
```

#### 이미 만들어진 리스트나 딕셔너리리를 언팩킹해서 전달할 수도 있다.

```python
# 리스트 언팩킹
def introduce(name, age, location):
    print(f"이름 : {name}, 나이 : {age}, 지역 : {location}")

user = ["KIM", 25, "SEOUL"]

introduce(*user)
```

```python
# 딕셔너리 언팩킹
def introduce(name, age, location):
    print(f"이름 : {name}, 나이 : {age}, 지역 : {location}")

user = {
    "name" : "KIM",
    "age" : 25,
    "location" : "SEOUL"
}

introduce(**user)
```
