#### 1. 파일 읽기(File Read)
프로그램에서 데이터를 사용하는 방법은 크게 두가지가 있다.
1. 메모리에 할당된 데이터를 읽기
2. 파일을 읽어서 해당 파일에 저장된 데이터를 읽기

메모리에 저장된 데이터는 프로그램이 종료되면 사라지기 떄문에, 데이터를 파일에 저장하면 프로그램을 다시 실행해도 데이터를 불러올 수 있다.

```python
# open() : 파일을 열 때 사용하는 함수
# r : 읽기
# w : 쓰기
# a : 이어쓰기
# x : 새 파일 생성

file = open("example.txt", "r")
```

```python
# read() : 파일의 전체 내용을 읽어오는 함수
file = open("example.txt", "r")
content = file.read()
print(content)
file.close()
```

```python
# readline() : 파일에서 한줄씩 읽어오는 함수
# readline()은 호출할때마다 다음줄로 이동한다.

file = open("example.txt", "r")
content1 = file.readline()
content2 = file.readline()
content3 = file.readline()
file.close()

print(content1)
print(content2)
print(content3)
```

```python
# readlines() : 파일의 모든 줄을 리스트 형태로 가져온다.
file = open("example.txt", "r")
content = file.readlines()
file.close()
print(content)
```

#### 실습
1. example.txt 파일을 읽기 모드로 열어라.
2. read()를 사용해서 전체 내용을 출력하라.
3. readline()을 사용해서 첫 번째 줄과 두 번째 줄을 출력하라.
4. readlines()를 사용해서 모든 줄을 리스트로 저장하라.
5. 리스트를 반복문으로 순회하면서 각 줄을 출력하라.

```python
file = open("example.txt", "r")
content = file.read()
print(content)
file.close()

file = open("example.txt", "r")
first_content = file.readline()
second_content = file.readline()
print(first_content)
print(second_content)
file.close()

file = open("example.txt", "r")
all_content = file.readlines()
print(all_content)
file.close()

for content in all_content:
    print(content)
```
