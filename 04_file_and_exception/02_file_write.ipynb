#### 1. 파일 쓰기(File Write)
파일 쓰기는 프로그램에서 생성한 파일이나 데이터를 저장할 수 있게 해준다.


```python
# write() : 문자열을 파일에 기록하는 함수
# 문자열만 기록 가능하기 때문에 int, bool등 다른 자료형이 오게 되면 오류가 발생한다.

# w : 기존 내용 삭제 후 새로 작성
# a : 기존 내용 유지 후 이어쓰기
# x : 파일이 이미 있으면 오류

file = open("output.txt", "w")
file.write("Hello Python")
file.close()
```

```python
# 이어쓰기 모드일 때 자동 줄바꿈을 지원하지 않기 때문에 줄바꿈을 추가해주어야 한다.
file = open("output.txt", "a")
file.write("\n")
file.write("Hello Java")
file.write("\n")
file.write("Hello Mysql")
file.close()
```

```python
# writelines() : 여러 문자열을 한번에 기록하는 함수
# 이 경우에도 자동 줄바꿈 지원은 하지 않기때문에 추가해주어야 한다.
languages = [
    "Python\n",
    "Java\n",
    "Mysql"
    ]

file = open("output.txt", "w")
file.writelines(languages)
file.close()
```
