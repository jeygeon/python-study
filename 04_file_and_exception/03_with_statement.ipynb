#### with란
`with`는 특정 작업이 끝났을 때 자원을 자동으로 정리해주는 문법이다.

`with`를 사용하면 `file.close()`를 직접해주지 않아도 자동으로 자원이 정리된다.

구조는 아래와 같다.
```python
with open("파일명", "모드") as 변수:
    실행할 코드
```

들여쓰기 된 코드 블록이 종료되면 자동으로 파일이 닫힌다.

````python
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
````

````python
# read() 뿐만 아니라 write(), readlines() 등 파일 읽기, 쓰기 함수는 모두 사용 가능하다.

with open("example.txt", "w") as file:
    file.write("hello python too")
````
