#### Exception Handing (예외처리)

코드가 실행되다가 오류가 발행하면 정상적인 실행 흐름이 중단된다. 때문에 예외처리를 해주는 것이 중요하다.

```python
print("시작")
print(10 / 0)
print("종료")
```

위와 같은 경우에는 `ZeroDivisionError` 에러가 발생하며 `print("종료")`가 실행되지 않았다.

이 경우 `try - except`를 사용하면 처리가 가능하다. `try` 안에서 에러가 발생하면 `except`로 이동하며 해당 코드가 실행이 되고 이후 코드는 정상적으로 동작하게 된다.

```python
print("시작")

try:
    print(10/0)

except:
    print("오류가 발생했습니다.")

print("종료")
```

```python
# 어떤 예외 타입이 발생했는지에 따라서 처리할 예외를 다르게 할 수도 있다.

print("시작")

try:
    number = int(input("숫자를 입력하세요: "))
    result = 100 / number

except ZeroDivisionError:
    print("0으로 나눌수는 없습니다.")

except ValueError:
    print("숫자만 입력이 가능합니다.")

print("종료")
```

```python
# 예외들을 튜플로 묶어서 하나로 처리할 수 있다.

print("시작")

try:
    number = int(input("숫자를 입력하세요: "))
    result = 100 / number

except (ZeroDivisionError, ValueError):
    print("정상적인 값을 입력하세요.")

print("종료")
```

```python
# Exception 객체를 사용하면 어떤 에러가 발생해는지 확인도 가능하다.

print("시작")

try:
    number = int(input("숫자를 입력하세요: "))
    result = 100 / number

except Exception as e:
    print(f"오류 발생 : {e}")
    print(type(e).__name__)

print("종료")
```

```python
# else는 예외가 발생하지 않았을 때 실행된다.

print("시작")

try:
    number = int(input("숫자를 입력하세요: "))
    result = 100 / number

except Exception as e:
    print(f"오류 발생 : {e}")
    print(type(e).__name__)

else:
    print(f"입력받은 숫자 : {number}")

print("종료")
```

```python
# finally를 사용하면 오류가 발생했던 안했던 반드시 실행된다.

print("시작")

try:
    number = int(input("숫자를 입력하세요: "))
    result = 100 / number

except Exception as e:
    print(f"오류 발생 : {e}")
    print(type(e).__name__)

else:
    print(f"입력받은 숫자 : {number}")

finally:
    print("종료")
```
