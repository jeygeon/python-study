#### Custom Exception (사용자 정의 예외)
사용자 정의 예외란 `ValueError`, `TypeError`, `ZeroDivisionError` 등 파이썬의 다양한 기본 예외말고

`InvalidPasswordError`와 같은 개발자가 직접 설정한 커스텀 예외를 말한다.

```python
# raise를 통해서 의도적으로 예외를 발생시킬 수 있다.
age = -10
if age < 0:
    raise ValueError
```

```python
# 파이썬에서는 Exception 클래스를 상속받아서 예외 클래스를 만들 수 있다.
# raise를 통해서 개발자가 의도한 상황에 직접 만든 커스텀 에러를 발생시킬 수 있다.
class InvalidAgeError(Exception):
    pass

try:
    age = -10

    if age < 0:
        raise InvalidAgeError

except InvalidAgeError:
    print("나이는 0보다 작을 수 없습니다.")
```

```python
# 또한 사용자 정의 예외에도 __init()__을 정의할 수 있다.
class InvalidAgeError(Exception):

    def __init__(self, age):
        self.age = age
        super().__init__(f"잘못된 나이입니다: {age}")

try:
    age = -10

    if age < 0:
        raise InvalidAgeError(age)

except InvalidAgeError as e:
    print(e)
    print(e.age)
```
