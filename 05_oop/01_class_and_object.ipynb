#### class and object (클래스와 객체)

#### 1. 객체지향 프로그래밍이란?
객체지향 프로그래밍은 프로그램을 여러개의 객체로 구성하고, 객체들이 서로 상호작용하도록 설계한 프로그래밍이다.

#### 2. class (클래스)란?
클래스는 객체 하나하나의 설계도이다.

```python
# class : 클래스를 선언하는 키워드
# User : 클래스 이름
# pass : 아직 구현할 내용이 없다는 의미

# 보통 클래스의 이름은 PascalCase를 사용한다.
# ex) UserAccount, MailService 등..

class User:
    pass
```

#### 3. object (오브젝트)린?
`object`는 클래스를 기반으로 생성된 객체 그 자체이다.

```python
class User:
    pass

user1 = User() # user1 객체
user2 = User() # user2 객체

# 전혀 다른 두 객체가 생성된 것을 확인할 수 있다.
print(user1 is user2) # False
```

#### 4. Attribute (속성)
객체는 데이터를 가질 수 있는데 이를 `Attribute(속성)` 이라고 한다.


```python
class User:
    pass

# 여기서 이름과 나이 지역을 속성으로 가졌다.
user1 = User()
user1.name = "KIM"
user1.age = 25
user1.location = "SEOUL"

print(user1.name, user1.age, user1.location)
```

#### 5. Method (메소드)
객체는 스스로도 기능을 수행할 수 있다. 이를 메소드라고 한다.

```python
class User:

    def study(self):
        print("공부중입니다.")

    def sleep(self):
        print("잠자고 있습니다.")

user = User()

user.study()
user.sleep()
```
