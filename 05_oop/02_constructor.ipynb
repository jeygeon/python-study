#### Constructor(생성자)
생성자란 객체가 생성될 때 자동으로 호출되는 메소드이다.

python에서는 `__init__()`이라는 메소드를 사용한다.

```python
class User():

    def __init__(self):
        print("User 객체가 생성되었습니다.")

user = User() # 이 때 객체가 생성되면서 __init__() 메소드를 실행
```

```python
# __init__()이 객체가 생성될 때 자동으로 실행되는 생성자이면, 이를 통해서 객체 생성하면서 매개변수 전달 또한 가능하다.
# self : 객체가 생성되는 자기 자신을 의미한다. 생성자에 무조건 첫번째로 와야하는 매개변수이다.

class User():

    def __init__(self, name, age):
        self.name = name
        self.age = age
        print(f"이름 : {name}, 나이 : {age} 객체가 생성되었습니다.")

user = User("KIM", 25) # 객체를 생성하면서 이름과 나이를 전달
```

```python
# 생성자의 매개변수는 처음 디폴트값 지정도 가능해서 아래와 같이 코드 작성도 가능하다.

class User():

    def __init__(self, name, level = "F"): # 최초 level을 F로 지정
        self.name = name
        self.level = level
        print(f"이름 : {name}, 레벨 : {level}의 객체가 생성되었습니다.")

    def change_level(self, level):
        before_level = self.level
        self.level = level
        print(f"레벨이 {before_level} 에서 {self.level}로 변경되었습니다.")

user = User("KIM")
user.change_level("B")
```
