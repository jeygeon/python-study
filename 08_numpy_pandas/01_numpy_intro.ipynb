#### numpy란
python에서 대규모 수치 데이터를 효율적으로 처리하기 위한 라이브러리이다.

일반적인 python 리스트 자료형으로도 데이터를 저장하고 다루는 것이 가능하지만, `numpy`를 사용하면 대량의 숫자 데이터를 다룰 때 좀더 효율적으로 다룰 수 있다.


주요한 특징
- 다차원 배열 지원
- 빠른 수치 연산
- 행렬 및 선형대수 연산
- 통계 및 수학 함수 제공
- Pandas, Scikit-learn 등 데이터 분석 라이브러리의 기반

```python
# numpy 불러오기
# np라는 별칭을 사용해서 불러올 수 있다.
import numpy as np

# numpy 버전 체크
print(np.__version__)
```

```python
numbers = np.array([1, 2, 3, 4, 5])
print(type(numbers)) # ndarray는 numpy에서 사용하는 다차원 배열 객체이다.
```

#### numpy에서는 배열에 연산을 적용할 수 있다.

```python
numbers = np.array([1, 2, 3, 4, 5])

print(numbers + 1) # 2 3 4 5 6
print(numbers - 1) # 0 1 2 3 4
print(numbers * 2) # 2  4  6  8 10
print(numbers / 2) # 0.5 1.  1.5 2.  2.5
```

#### numpy 배열에는 데이터를 파악할 수 있는 다양한 속성이 있다.

```python
numbers = np.array([1, 2, 3, 4, 5])

# 배열의 차원
print(numbers.ndim)

# 배열의 형태
print(numbers.shape) # (5,) 은 1차원의 배열에 원소가 5개 있다는 뜻이다.

# 전체 원소 갯수
print(numbers.size)

# 원소의 데이터 타입
print(numbers.dtype)
```

#### Numpy 기초 실습

`scores = np.array([85, 90, 78, 92, 88])`

요구사항:
1. 전체 점수의 합계를 구한다.
2. 평균을 구한다.
3. 최고 점수를 구한다.
4. 최저 점수를 구한다.
5. 모든 학생의 점수에 보너스 5점을 더한다.
6. 보너스 적용 후 평균을 구한다.

```python
scores = np.array([85, 90, 78, 92, 88])

# 1. 전체 점수의 합계를 구한다.
total_score = scores.sum()
print(total_score)

# 2.평균을 구한다.
mean_score = scores.mean()
print(mean_score)

# 3. 최고 점수를 구한다.
max_score = scores.max()
print(max_score)

# 4. 최젖 점수를 구한다.
min_score = scores.min()
print(min_score)

# 5. 모든 학생의 점수에 보너스 5점을 더한다.
add_score = scores + 5
print(add_score)

# 6. 보너스 적용후 평균을 구한다.
print(add_score.mean())
```
