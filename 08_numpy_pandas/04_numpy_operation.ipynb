#### 1️⃣ 배열간 산술 연산
`numpy`에서는 배열간의 연산을 수행할 수 있다.


```python
import numpy as np
```

```python
# 각 배열의 같은 위치의 원소끼리 연산을 수행한다.

arr1 = np.array([10, 20, 30])
arr2 = np.array([1, 2, 3])

print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)

print(arr1 > 10)
print(arr1 == 10)
print(arr1 < 10)
```

#### 2️⃣ 집계 함수
배열의 데이터를 요약하는 함수

```python
scores = np.array([80, 90, 35, 60, 50])

print(scores.sum())  # 합
print(scores.mean()) # 평균
print(scores.max())  # 최대값
print(scores.min())  # 최솟값
print(scores.std())  # 표준편차
print(scores.var())  # 분산
```

#### 3️⃣ axis를 활용한 집계


```python
scores = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

# 행끼리 합
print(scores.sum(axis = 1))

# 열끼리 합
print(scores.sum(axis = 0))

# 행끼리 평균
print(scores.mean(axis = 1))

# 열끼리 평균
print(scores.mean(axis = 0))
```

#### 4️⃣ 정렬 (sorting)
- `np.sort(배열)` : 배열의 내용은 건들지 않고 새로운 정렬된 배열 반환
- `배열.sort()` : 새로운 배열 반환없이 기존 배열을 정렬함

```python
arr = np.array([50, 30, 25, 60, 90])

# 오름차순 정렬
asc_arr = np.sort(arr)
print(asc_arr)

# 내림차순 정렬
desc_arr = np.sort(arr)[::-1]
print(desc_arr)

# 오름차순 정렬
arr.sort()
print(arr)

# 내림차순 정렬
print(arr[::-1])
```

#### 5️⃣ 배열 합치기와 나누기
- `concatenate()` : 두 배열을 합칠 수 있다.
- `split()` : 두 배열을 나눌 수 있다.

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

result = np.concatenate((arr1, arr2))
print(result)
```

```python
# 2차원의 배열은 asix로 합치는 방향을 지정할 수 있다.
arr1 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

arr2 = np.array([
    [7, 8, 9],
    [10, 11, 12]
])

result1 = np.concatenate((arr1, arr2), axis=1)
print(result1)

result2 = np.concatenate((arr1, arr2), axis=0)
print(result2)
```

```python
# split()

arr = np.array([1, 2, 3, 4, 5, 6])
result = np.split(arr, 2)
print(result)
```
