#### Numpy Broadcasting
서로 다른 형태의 배열끼리 연산을 가능하게 `numpy`가 배열을 맞춰주는 기능이다.

```python
import numpy as np
```

```python
arr = np.array([1, 2, 3])
result = arr + 5
print(result)
```

위와 같은 연산은 내부적으로 다음과 같이 처리한다.<br/>
`[1, 2, 3] + [5, 5, 5]`

숫자 5를 배열의 크게 맞춰서 확장을 한다.


```python
# 1차원 배열의 브로드캐스팅
arr1 = np.array([1, 2, 3])
arr2 = np.array([3])

print(arr1 + arr2)
```
