#### Pandas란?
`pandas`란 파이썬에서 데이터를 분석하고 가공하기 위한 라이브러리이다.

```python
import pandas as pd

# pandas 버전 확인
print(pd.__version__)
```

`pandas`는 크게 두가지 자료구조가 있다.
1. `Series` : 인덱스와 값으로 구성된 1차원 데이터 구조
2. `DataFrame` : 행과 열로 구성된 2차원 데이터 구조

```python
import pandas as pd
import numpy as np

data = {
    "name": ["Kim", "Lee", "Park", "Choi"],
    "age": [25, 30, 28, 35],
    "salary": [3000, 4000, 3500, 5000]
}

# 1. data를 DataFrame으로 변환하고 df에 저장하기
df = pd.DataFrame(data)

# 2. df 출력하기
print(df)

# 3. Pandas 버전 출력하기
print(pd.__version__)

# 4. df의 자료형 확인하기 (type() 사용)
print(type(df))
```
