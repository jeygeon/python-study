#### 💡 series란?
`series`란 `pandas`에서 사용하는 1차원 데이터 구조이다.<br/>
쉽게 말하면 인덱스(index)와 값(value)로 구성된 데이터이다.

```python
import pandas as pd
```

```python
scores = pd.Series([90, 85, 30, 60, 75])
print(scores)

# 왼쪽 숫자 : index
# 오른쪽 숫자 : value
# dtype : 데이터의 자료형
# 0    90
# 1    85
# 2    30
# 3    60
# 4    75
# dtype: int64
```

`numpy`에는 인덱스가 따로 지정돼서 생성되지는 않지만 `pandas`는 기본 인덱스 0부터 자동생성된다.

```python
# pandas는 인덱스를 직접 지정해서 표현할 수도 있다.
scores = pd.Series(
    [90, 50, 75],
    index=["KIM", "LEE", "PARK"]
    )

print(scores)
```

```python
# 딕셔너리를 사용해서 표현 할 수도 있다. 이 경우에는 key값이 인덱스가 된다.
scores = pd.Series({
    "KIM" : 90,
    "LEE" : 50,
    "PARK" : 75
})

print(scores)
```

```python
# series의 속성 확인
scores = pd.Series({
    "KIM" : 90,
    "LEE" : 50,
    "PARK" : 75
})

# 인덱스
print(scores.index)

# 값
print(scores.values)

# 차원
print(scores.ndim)

# 형태
print(scores.shape)

# 자료형
print(scores.dtype)

# 원소 갯수
print(scores.size)
```

```python
# series는 인덱스를 통해서 데이터 값을 가져올 수 있다.
scores = pd.Series({
    "KIM" : 90,
    "LEE" : 50,
    "PARK" : 75
})

print(scores["KIM"])
print(scores["PARK"])
print(scores[["KIM", "LEE"]])

# 슬라이싱을 통해서도 가져올 수 있다 (이 경우에는 리스트처럼 마지막 인덱스가 포함되지 않는다.)
print(scores[0:2])
```

#### loc와 iloc
- `loc` : 인덱스의 이름(Label)을 기준으로 접근
- `iloc` : 인덱스의 위치(Position)를 기준으로 접근
<br/><br/>
> 슬라이싱을 할 때 `loc`는 마지막을 포함하지만 `iloc`는 마지막을 포함하지 않는다

```python
scores = pd.Series({
    "KIM" : 90,
    "LEE" : 50,
    "PARK" : 75
})

# loc
print(scores.loc["KIM"])

# iloc
print(scores.iloc[2])

# 마지막 포함
print(scores.loc["KIM" : "PARK"])

# 마지막 포함하지 않음
print(scores.iloc[0:2])
```

```python
# series도 numpy처럼 산술 연산이 가능하다.
scores = pd.Series([70, 85, 55])

print(scores + 10)
print(scores - 20)
print(scores.sum())
print(scores.mean())
print(scores.max())
print(scores.min())
```

#### 실습문제

1. scores 출력하기
2. Series의 인덱스 출력하기
3. Series의 값 출력하기
4. Series의 차원과 shape 출력하기
5. Kim의 점수 출력하기 (loc 사용)
6. 세 번째 학생의 점수 출력하기 (iloc 사용)
7. Kim부터 Park까지의 점수 출력하기 (loc 사용)
8. 모든 점수에 5점 더하기

```python
scores = pd.Series(
    [90, 80, 75, 95, 85],
    index=["Kim", "Lee", "Park", "Choi", "Jung"]
)

# 1. scores 출력하기
print(scores)

# 2. Series의 인덱스 출력하기
print(scores.index)

# 3. Series의 값 출력하기
print(scores.values)

# 4. Series의 차원과 shape 출력하기
print(scores.ndim)
print(scores.shape)

# 5. Kim의 점수 출력하기(loc사용)
print(scores.loc["Kim"])

# 6. 세 번째 학생의 점수 출력하기(iloc사용)
print(scores.iloc[2])

# 7. Kim부터 Park까지 점수 출력하기(loc사용)
print(scores.loc["Kim" : "Park"])

# 8. 모든 점수에 5점 더하기
print(scores + 5)
```
