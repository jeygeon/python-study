#### DataFrame이란
행(row)과 열(column)으로 구성된 2차원 데이터 구조이다.

```python
import pandas as pd
```

```python
# 딕셔너리로 사용하기
# key가 열 이름이 되고 value가 행이 된다.

data = {
    "name": ["Kim", "Lee", "Park"],
    "age": [25, 30, 28],
    "score": [90, 80, 75]
}

df = pd.DataFrame(data)
print(df)
```

```python
# 리스트로 사용하기
# columns= 로 각 컬럼에 부여할 이름을 정해줘야한다.

data = [
    ["Kim", 25, 90],
    ["Lee", 30, 80],
    ["Park", 28, 75]
]

df = pd.DataFrame(data, columns=["name", "age", "score"])
print(df)
```

```python
# 인덱스 직접 지정하기

data = {
    "name": ["Kim", "Lee", "Park"],
    "age": [25, 30, 28],
    "score": [90, 80, 75]
}

df = pd.DataFrame(data, index=["student1", "student2", "student3"])
print(df)
```

#### 데이터 미리보기
- `head()` : 앞에서부터 데이터 확인 (디폴트 : 5)
- `tail()` : 뒤에서부터 데이터 확인 (디폴트 : 5)
- `info()` : Dataframe의 전체적인 정보 확인
- `describe()` : 숫자데이터에 기초 통계랑을 확인

```python
data = {
    "name": ["Kim", "Lee", "Park"],
    "age": [25, 30, 28],
    "score": [90, 80, 75]
}

df = pd.DataFrame(data, index=["student1", "student2", "student3"])

print(df.head())
print("======================================")
print(df.tail())
print("======================================")
print(df.info())
print("======================================")
print(df.describe())
```

#### 실습문제
1. data를 DataFrame으로 변환해서 df에 저장하기
2. DataFrame 전체 출력하기
3. DataFrame의 행 인덱스 확인하기
4. DataFrame의 열 이름 확인하기
5. DataFrame의 행과 열 개수 확인하기
6. DataFrame의 각 열 자료형 확인하기
7. 앞에서 3개 행 출력하기 (head() 사용)
8. 뒤에서 2개 행 출력하기 (tail() 사용)
9. DataFrame의 기본 정보 확인하기 (info() 사용)
10. 숫자형 데이터의 기초 통계량 확인하기 (describe() 사용)

```python
data = {
    "name": ["Kim", "Lee", "Park", "Choi", "Jung"],
    "age": [25, 30, 28, 35, 27],
    "department": ["IT", "HR", "IT", "Sales", "HR"],
    "salary": [3000, 4000, 3500, 5000, 3800]
}

# 1. data를 DataFrame으로 변환해서 df에 저장하기
df = pd.DataFrame(data)

# 2. DataFrame 전체 출력하기
print(df)

# 3. DataFrame의 행 인덱스 확인하기
print(df.index)

# 4. DataFrame의 열 이름 확인하기
print(df.columns)

# 5. DataFrame의 행과 열 개수 확인하기
print(df.shape)

# 6. DataFrame의 각 열 자료형 확인하기
print(df.dtypes)

# 7. 앞에서 3개 행 출력하기 (head() 사용)
print(df.head(3))

# 8. 뒤에서 2개 행 출력하기 (tail() 사용)
print(df.tail(2))

# 9. DataFrame의 기본 정보 확인하기 (info() 사용)
print(df.info())

# 10. 숫자형 데이터의 기초 통계량 확인하기 (describe() 사용)
print(df.describe())
```
