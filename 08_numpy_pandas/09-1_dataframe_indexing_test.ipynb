#### 💡 연습문제 1

```python
import pandas as pd
```

```python
data = {
    "order_id": [1001, 1002, 1003, 1004, 1005, 1006],
    "customer": ["Kim", "Lee", "Park", "Choi", "Jung", "Han"],
    "product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Mouse", "Laptop"],
    "price": [1500000, 30000, 80000, 350000, 30000, 1400000],
    "quantity": [1, 2, 1, 1, 3, 2],
    "status": ["배송완료", "배송중", "배송완료", "주문취소", "배송중", "배송완료"]
}

df = pd.DataFrame(data)
print(df)
```

```python
# A. 행과 열 선택
# 1. customer 열 전체를 출력
print(df["customer"])

# 2. customer와 product 열만 출력
print(df[["customer", "product"]])

# 3. loc을 사용해서 인덱스가 2인 행을 출력
print(df.loc[2])

# 4. iloc을 사용해서 네 번째 행을 출력
print(df.iloc[3])

# 5. loc을 사용해서 인덱스 1부터 4까지의 order_id, product 열을 출력
print(df.loc[1:4, ["order_id", "product"]])

# 6. iloc을 사용해서 첫 번째 행부터 세 번째 행까지, 첫 번째 열부터 네 번째 열까지 출력
print(df.iloc[0:3, 0:4])
```

```python
# B. 조건으로 데이터 검색
# 7. 가격이 100,000원 이상인 주문을 출력
print(df[df["price"] >= 100000])

# 8. 상품이 Laptop인 주문만 출력
print(df[df["product"] == "Laptop"])

# 9. 배송 상태가 배송중인 주문만 출력
print(df[df["status"] == "배송중"])

# 10. 가격이 50,000원 이상이면서 수량이 2개 이상인 주문을 출력
print(df[(df["price"] >= 50000) & (df["quantity"] >= 2)])

# 11. 상품이 Laptop이거나 가격이 300,000원 이상인 주문을 출력
print(df[(df["product"] == "Laptop") | (df["price"] >= 300000)])
```

```python
# C. 데이터 수정
# 12. 인덱스 0번 행의 상품 가격을 1,600,000원으로 수정
df.loc[0, "price"] = 1600000

# 13. 상품이 Mouse인 주문의 가격을 모두 35,000원으로 수정
df.loc[df["product"] == "Mouse", "price"] = 35000

# 14. 배송 상태가 배송중인 주문의 수량을 모두 1개씩 증가
df.loc[df["status"] == "배송중", "quantity"] += 1

# 15. 가격이 1,000,000원 이상인 주문의 배송 상태를 고가상품으로 변경
df.loc[df["price"] >= 1000000, "status"] = "고가상품"
```

#### 💡 연슴문제 2


```python
data = {
    "customer": ["Kim", "Lee", "Park", "Choi", "Jung", "Han", "Kang", "Yoon"],
    "category": ["전자제품", "의류", "전자제품", "식품", "의류", "식품", "전자제품", "의류"],
    "product": ["Laptop", "T-Shirt", "Monitor", "Apple", "Jeans", "Coffee", "Keyboard", "Jacket"],
    "price": [1800000, 35000, 450000, 5000, 80000, 25000, 120000, 150000],
    "quantity": [1, 3, 2, 10, 2, 5, 1, 2],
    "status": ["배송완료", "배송중", "배송중", "배송완료", "주문취소", "배송중", "배송완료", "배송중"]
}

df = pd.DataFrame(
    data,
    index=["ORD001", "ORD002", "ORD003", "ORD004",
           "ORD005", "ORD006", "ORD007", "ORD008"]
)

print(df)
```

```python
# A. 행과 열 선택
# 1. customer와 product 열만 출력
print(df[["customer", "product"]])

# 2. ORD003 주문의 모든 정보를 출력(loc 사용)
print(df.loc["ORD003"])

# 3. iloc을 사용해서 다섯 번째 행의 모든 정보를 출력
print(df.iloc[4])

# 4. ORD002부터 ORD006까지의 행에서 customer, price, status 열만 출력
print(df.loc["ORD002" : "ORD006", ["customer", "price", "status"]])

# 5. iloc을 사용해서 두 번째 행부터 여섯 번째 행까지, 세 번째 열부터 다섯 번째 열까지 출력
print(df.iloc[1:6, 2:5])
```

```python
# B. 복합 조건 검색
# 6. 가격이 100,000원 이상이면서 수량이 2개 이상인 주문을 출력
print(df[(df["price"] >= 100000) & (df["quantity"] >= 2)])

# 7. 카테고리가 전자제품이면서 배송 상태가 배송완료인 주문을 출력
print(df[(df["category"] == "전자제품") & (df["status"] == "배송완료")])

# 8. 가격이 50,000원 이하이거나 수량이 5개 이상인 주문을 출력
print(df[(df["price"] <= 50000) | (df["quantity"] >= 5)])

# 9. 배송 상태가 주문취소가 아닌 주문만 출력
print(df[df["status"] != "주문취소"])

# 10. 카테고리가 의류 또는 전자제품이면서 가격이 100,000원 이상인 주문을 출력
print(df[(
    (df["category"] == "의류") |
    (df["category"] == "전자제품")
) & (df["price"] >= 100000)])

# 11. 고객이 Kim이 아니면서 배송 상태가 배송중인 주문을 출력
print(df[(df["customer"] != "Kim") & (df["status"] == "배송중")])
```

```python
# C. 조건에 따른 데이터 수정
# 12. ORD001 주문의 수량을 2개로 변경
df.loc["ORD001", "quantity"] = 2

# 13. 카테고리가 의류인 모든 주문의 가격을 10% 인상
df.loc[df["category"] == "의류", "price"] *= 1.1

# 14. 배송 상태가 배송중이면서 수량이 3개 이상인 주문의 상태를 우선배송으로 변경
df.loc[((df["status"] == "배송중") & (df["quantity"] >= 3)), "status"] = "우선배송"

# 15. 가격이 100,000원 이상인 모든 주문에 대해 가격을 50,000원씩 할인
df.loc[df["price"] >= 100000, "price"] -= 50000

# 16. 카테고리가 식품인 주문의 배송 상태를 모두 식품배송으로 변경
df.loc[df["category"] == "식품", "status"] = "식품배송"
```

```python
# 17. 가격이 100,000원 이상이고 수량이 2개 이상인 주문만 선택해서 price 열을 출력
print(df.loc[(df["price"] >= 100000) & (df["quantity"] >= 2), "price"])

# 18. 배송완료 상태가 아닌 주문 중 가격이 100,000원 이상인 주문의 가격을 5% 인하
df.loc[
    (df["status"] != "배송완료") & (df["price"] >= 100000),
    "price"
] = (
    df.loc[
        (df["status"] != "배송완료") & (df["price"] >= 100000),
        "price"
    ] * 0.95
).round().astype(int)

# 19. 전자제품 카테고리의 주문 중 가격이 가장 높은 주문의 모든 정보를 출력
max_price = df.loc[df["category"] == "전자제품", "price"].max()
print(df.loc[(df["category"] == "전자제품") & (df["price"] == max_price)])
```
