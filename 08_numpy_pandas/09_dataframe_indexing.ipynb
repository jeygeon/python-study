#### 💡 DataFrame Indexing & Slicing

````python
import pandas as pd

data = {
    "name": ["Kim", "Lee", "Park", "Choi", "Jung"],
    "age": [25, 30, 28, 35, 27],
    "department": ["IT", "HR", "IT", "Sales", "HR"],
    "salary": [3000, 4000, 3500, 5000, 3800]
}

df = pd.DataFrame(data)

print(df)
````

````python
# 열 선택하기
print(df["name"])

# 선택된 열의 타입은 Series 타입인 것을 확인할 수 있다.
print(type(df["name"]))
````

````python
# 여러 열 선택도 가능하다. (리스트로 전달)
print(df[["name", "age", "department"]])

# 여러 열을 선택했을 때 데이터 타입은 DataFrame이다.
print(type(df[["name", "age", "department"]]))
````

````python
# loc는 행의 인덱스를 기준으로 데이터를 가져올 수 있다.
print(df.loc[0])

# 여러개도 가능하다.
# 리스트로 전달
print(df.loc[[0, 1, 2]])

# 슬라이싱으로 전달 (주의 : 끝 인덱스도 포함한다)
print(df.loc[0:2])
````

````python
# iloc는 행의 위치로 데이터를 가져온다.
print(df.iloc[0])

# 슬라이싱도 가능하다. (주의 : 마지막은 포함되지 않는다.)
print(df.iloc[1:3])
````

````python
# 행과 열을 동시에 가져오기
# loc[행, 열], iloc[행, 열]
print(df.loc[1, "name"])
print(df.loc[3, "age"])

print(df.iloc[2, 1])
print(df.iloc[0, 0])
````

````python
# 조건으로 데이터 가져오기
print(df[df["age"] >= 30])

# 여러개 조건으로 데이터 가져오기
# 조건이 여러개로 가져올 경우 각 조건을 괄호로 감싸야 한다.
print(df[(df["age"] >= 30) & (df["salary"] >= 4000)])
````

````python
# 데이터 수정하기
print(df.loc[0, "age"])

df.loc[0, "age"] = 30
print(df.loc[0, "age"])
````

````python
# 여러 행 수정도 가능하다.
print(df[df["department"] == "IT"])

df.loc[df["department"] == "IT", "salary"] += 100

print(df[df["department"] == "IT"])
````
