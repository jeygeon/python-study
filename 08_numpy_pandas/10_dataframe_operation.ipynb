#### DataFrame의 산술연산

```python
import pandas as pd

data = {
    "name": ["Kim", "Lee", "Park", "Choi", "Jung"],
    "department": ["IT", "HR", "IT", "Marketing", "HR"],
    "salary": [3000, 3500, 4000, 3200, 3800],
    "bonus": [300, 200, 500, 100, 400]
}

df = pd.DataFrame(data)

print(df)
```

```python
# 직원별 총 급여 계산
# 새로운 컬럼이 추가된다
df["total_salary"] = df["salary"] + df["bonus"]
```

```python
# 다양한 산술 연산 가능
# 계산만 하고 변수나 열에 대입하지 않으면 원본 DataFrame은 바뀌지 않는다.

# 급여 10% 인상
df["salary"] * 1.1

# 급여 500 차감
df["salary"] - 500

# 급여를 12개월로 나누기
df["salary"] / 12

# 보너스 2배
df["bonus"] * 2
```

```python
df["annual_salary"] = df["salary"] * 1.1
print(df)
```

#### 실습문제


```python
# 1. 모든 직원의 salary에 500을 더한 결과 출력
print(df["salary"] + 500)

# 2. 모든 직원의 bonus를 2배로 계산한 결과 출력
print(df["bonus"] * 2)

# 3. salary와 bonus를 더한 total_salary 열 생성
df["total_salary"] = df["salary"] + df["bonus"]

# 4. salary의 10%를 계산한 tax 열 생성
df["tax"] = df["salary"] * 0.1

# 5. total_salary가 4000 이상인 직원만 출력
print(df[df["total_salary"] >= 4000])
```
