#### 결측치란
결측치(Missing Data)란 데이터가 존재해야하는 위치에 값이 없는 상태를 말한다.

```python
import pandas as pd
import numpy as np

# 결측치가 확인되면 정수형 데이터가 실수형 데이터로 변경된다.

data = {
    "name": ["Kim", "Lee", "Park", "Choi", "Jung"],
    "age": [25, 30, np.nan, 32, 27],
    "department": ["IT", "HR", "IT", np.nan, "HR"],
    "salary": [3000, 3500, 4000, np.nan, 3800]
}

df = pd.DataFrame(data)

print(df)
```

```python
# 결측치 확인하기 : isnull()
print(df.isnull())

# 반대로 확인하기 : notnull()
print(df.notnull())
```

```python
# 컬럼별 결측치 갯수 확인
print(df.isnull().sum())

# 전체 결측치 갯수 확인 : sum() 한번 더
print(df.isnull().sum().sum())
```

```python
# 결측치가 아닌 데이터 가져오기
# age가 비어있지 않은 데이터만 가져오기
print(df[df["age"].notnull()])

# age가 비어있는 데이터만 가져오기
print(df[df["age"].isnull()])
```

```python
# 결측치가 하나라도 있는 행 삭제
df.dropna()

# 결측치가 하나라도 있는 열 삭제
df.dropna(axis=1)

# 특정 열에 결측치가 있는 행만 삭제
df.dropna(subset=["age"])

# 여러 열도 지정 가능
df.dropna(subset=["age", "department"])
```

```python
# 하나라도 결측치가 있는 행은 삭제
df.dropna(how="any")

# 모든 열이 결측치인행 삭제
df.dropna(how="all")
```

`dropna()`는 기본적으로 DataFrame의 값을 변경하지 않는다.

`new_df = df.dropna()`처럼 새로운 변수에 값을 저장하거나, `df.dropna(inplace=True)`처럼 `inplace=True`를 전달해야 원본에 반영된다.

#### 결측치 대체하기 : `fillna()`
결측치가 있는 행을 삭제하는 대신 다른 값으로 채우는 방법이다.

```python
# 특정 값으로 대체
df.fillna(0)
```

```python
# 특정 열만 변경
df["age"] = df["age"].fillna(0)
df["department"] = df["department"].fillna("미정")
df["salary"] = df["salary"].fillna(0)
print(df)
```

```python
# 평균으로 변경
df["age"] = df["age"].fillna(df["age"].mean())

# 중앙값으로 변경
df["salary"] = df["salary"].fillna(df["salary"].median())
```

#### 실습문제 3

```python
data = {
    "name": ["Kim", "Lee", "Park", "Choi", "Jung", "Han"],
    "age": [25, np.nan, 30, 28, np.nan, 35],
    "department": ["IT", "HR", np.nan, "IT", "HR", np.nan],
    "salary": [3000, 3500, 4000, np.nan, 3800, 4500],
    "bonus": [300, np.nan, 500, 200, np.nan, 600]
}

df = pd.DataFrame(data)
print(df)
```

```python
# 1. 각 열의 결측치 개수를 출력
print(df.isnull().sum())

# 2. age의 결측치를 age 중앙값으로 대체
df["age"] = df["age"].fillna(df["age"].median())

# 3. department의 결측치를 "미정"으로 대체
df["department"] = df["department"].fillna("미정")

# 4. salary의 결측치를 salary 평균으로 대체
df["salary"] = df["salary"].fillna(df["salary"].mean())

# 5. bonus의 결측치를 0으로 대체
df["bonus"] = df["bonus"].fillna(0)

# 6. 모든 결측치가 처리되었는지 확인
print(df.isnull().sum())
```
