#### groupby()란
`groupby()`란 데이터를 특정 컬럼의 값에 따라 그룹으로 나누는 기능이다.

```python
df.groupby("그룹 기준 컬럼")["계산할 컬럼"].집계함수()
```

````python
import pandas as pd
import numpy as np

data = {
    "name": ["Kim", "Lee", "Park", "Choi", "Jung",
             "Han", "Kang", "Yoon", "Lim", "Shin"],
    "department": ["IT", "HR", "IT", "Marketing", "HR",
                   "IT", "Marketing", "HR", "IT", "Marketing"],
    "position": ["Junior", "Senior", "Senior", "Junior", "Junior",
                 "Senior", "Senior", "Junior", "Junior", "Senior"],
    "age": [25, 35, 32, 27, 29, 38, 36, 26, 31, 34],
    "salary": [3000, 4500, 5000, 3200, 3500,
               5500, 4800, 3300, 3800, 5200],
    "bonus": [300, 500, 700, 200, 300,
              800, 600, 250, 400, 750]
}

df = pd.DataFrame(data)

print(df)
````

````python
# ex) 부서별 평균 급여 구하기

# groupby("department") : 부서별 그룹화
# ["salary"] : 급여 컬럼 선택
# .mean() : 평균 계산
df.groupby("department")["salary"].mean()
````

````python
# 부서별 급여 총합
df.groupby("department")["salary"].sum()

# 부서별 평균 급여
df.groupby("department")["salary"].mean()

# 부서별 최고 급여
df.groupby("department")["salary"].max()

# 부서별 최저 급여
df.groupby("department")["salary"].min()

# 부서별 직원 수
df.groupby("department").size()
````

#### 주의사항
- `count()` : 특정 컬럼에서 결측치를 제외한 갯수
- `size()` : 그룹에 속한 전체 행의 갯수

````python
# 두개 이상의 컬럼으로 그룹화 할 수도 있다.
df.groupby(["department", "position"])["salary"].mean()
````

````python
# agg()를 사용하면 여러 집계함수를 한번에 사용할 수 있다.
df.groupby("department")["salary"].agg(["mean", "max", "min"])
````

````python
# agg()를 사용해 서로 다른 컬럼에 적용할 수도 있다.
df.groupby("department").agg({
    "salary" : "mean",
    "bonus" : "max",
    "age" : "min"
})
````

#### 연습문제

````python
# 1. 부서별 평균 급여 확인
print(df.groupby("department")["salary"].mean())

# 2. 부서별 급여 총합 확인
print(df.groupby("department")["salary"].sum())

# 3. 부서별 최고 급여 확인
print(df.groupby("department")["salary"].max())

# 4. 부서별 최저 급여 확인
print(df.groupby("department")["salary"].min())

# 5. 부서별 직원 수 확인
print(df.groupby("department").size())

# 6. 부서별 평균 보너스 확인
print(df.groupby("department")["bonus"].mean())
````

````python
# 7. 부서와 직급별 평균 급여 확인
print(df.groupby(["department", "position"])["salary"].mean())

# 8. 직급별 평균 나이 확인
print(df.groupby("position")["age"].mean())

# 9. 부서와 직급별 직원 수 확인
print(df.groupby(["department", "position"]).size())

# 10. 부서별 평균 급여, 평균 보너스, 최고 나이를 한 번에 확인
print(df.groupby("department").agg({
    "salary" : "mean",
    "bonus" : "mean",
    "age" : "max"
}))
````

````python
# 11. 평균 급여가 가장 높은 부서 확인
print(df.groupby("department")["salary"].mean().idxmax())

# 12. 부서별 평균 급여를 높은 순서대로 정렬
print(df.groupby("department")["salary"].mean().sort_values(ascending=False))

# 13. 부서별 최고 급여를 받는 직원의 전체 정보 출력
idx = df.groupby("department")["salary"].idxmax()
print(df.loc[idx])
````
