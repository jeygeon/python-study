#### merge()
공통 컬럼을 기준으로 데이터를 결합하는 방식이다. (sql의 join같은)

```python
import pandas as pd

# 직원 정보
employees = pd.DataFrame({
    "emp_id": [1, 2, 3, 4, 5],
    "name": ["Kim", "Lee", "Park", "Choi", "Jung"],
    "department_id": [101, 102, 101, 103, 102],
    "salary": [3000, 4500, 5000, 3200, 3500]
})

# 부서 정보
departments = pd.DataFrame({
    "department_id": [101, 102, 103, 104],
    "department_name": ["IT", "HR", "Marketing", "Sales"]
})

print(employees)
print(departments)
```

```python
# inner join : 교집합만 결합
pd.merge(employees, departments, on="department_id", how="inner")
```

```python
# left join : 왼쪽 전체 유지
pd.merge(employees, departments, on="department_id", how="left")
```

```python
# right join : 오른쪽 전체유지
pd.merge(employees, departments, on="department_id", how="right")
```

```python
# outer join : 모든 데이터 프레임 전체 유지
pd.merge(employees, departments, on="department_id", how="outer")
```

#### concat()
`concat()`은 여러 DataFrame을 행 방향 또는 열 방향으로 이어붙히는 기능이다

```python
import pandas as pd

df1 = pd.DataFrame({
    "name": ["Kim", "Lee", "Park"],
    "salary": [3000, 4000, 5000]
})

df2 = pd.DataFrame({
    "name": ["Choi", "Jung", "Han"],
    "salary": [3500, 4500, 5500]
})

print(df1)
print(df2)
```

```python
# axis=0 : 행 이어붙히기 (디폴트)
# 단순 result 이어붙히기라서 인덱스가 변하지 않음
print(pd.concat([df1, df2], axis=0))
```

```python
# ignore_index=True : 새로운 인덱스 부여
print(pd.concat([df1, df2], ignore_index=True))
```

```python
# axis=1 : 열 방향 결합
df3 = pd.DataFrame({
    "name": ["Kim", "Lee", "Park"],
    "department": ["IT", "HR", "Marketing"]
})

df4 = pd.DataFrame({
    "age": [25, 30, 28],
    "salary": [3000, 4000, 5000]
})

result = pd.concat([df3, df4], axis=1)

print(result)
```

#### 실습문제

```python
# 1. employees와 departments를 INNER JOIN으로 결합
print(pd.merge(employees, departments, on="department_id", how="inner"))

# 2. employees를 기준으로 LEFT JOIN 실행
print(pd.merge(employees, departments, on="department_id", how="left"))

# 3. departments를 기준으로 RIGHT JOIN 실행
print(pd.merge(employees, departments, on="department_id", how="right"))

# 4. OUTER JOIN 실행
result = pd.merge(employees, departments, on="department_id", how="outer")

# 5. 결합한 결과에서 name, department_name, salary 컬럼만 출력
print(result[["name", "department_name", "salary"]])
```

```python
# 1. df1과 df2를 행 방향으로 결합
print(pd.concat([df1, df2], axis=0))

# 2. df1과 df2를 행 방향으로 결합하되 인덱스를 0부터 다시 생성
print(pd.concat([df1, df2], ignore_index=True))

# 3. df3과 df4를 열 방향으로 결합
print(pd.concat([df3, df4], axis=1))

# 4. df1, df2, df3을 한 번에 행 방향으로 결합
print(pd.concat([df1, df2, df3], axis=0))

# 5. df1과 df2를 결합한 결과를 result 변수에 저장하고,
#    salary의 평균을 출력
result = pd.concat([df1, df2], axis=0)
print(result.loc[:, "salary"].mean())
```
