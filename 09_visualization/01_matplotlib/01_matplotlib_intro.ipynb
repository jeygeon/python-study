#### Matplotlib란?
`Matplotlib`는 python에서 데이터를 그래프로 표현할 수 있도록 해주는 라이브러리이다.

```python
# 설치 및 불러오기
import matplotlib
print(matplotlib.__version__)

# 그래프를 그릴때는 보통 pyplot을 사용한다.
import matplotlib.pyplot as plt
```

```python
# 막대 그래프 그리기
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y)

# 그래프 제목
plt.title("My First Graph")
# x축 이름
plt.xlabel("X Axis")
# y축 이름
plt.ylabel("Y Axis")
# 그래프 출력
plt.show()
```

#### 실습문제 1

```python
# 1. matplotlib.pyplot을 plt라는 이름으로 불러오기
import matplotlib.pyplot as plt

# 2. x 데이터 생성: 1, 2, 3, 4, 5
x = [1, 2, 3, 4, 5]

# 3. y 데이터 생성: 5, 10, 15, 20, 25
y = [5, 10, 15, 20, 25]

# 4. x, y 데이터를 이용해 선 그래프 그리기
plt.plot(x, y)

# 5. 그래프 제목을 "My First Visualization"으로 설정
plt.title("My First Visualization")

# 6. x축 이름을 "Month", y축 이름을 "Sales"로 설정
plt.xlabel("Month")
plt.ylabel("Sales")

# 7. 그래프 출력
plt.show()
```

#### 실습문제 2

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
exercise_time = [30, 45, 0, 60, 50, 90, 40]

# matplotlib.pyplot을 plt로 import 하기
import matplotlib.pyplot as plt

# 요일과 운동 시간을 이용해 선 그래프 그리기
plt.plot(days, exercise_time)

# 그래프 제목을 Weekly Exercise Time으로 설정
plt.title("Weekly Exercise Time")

# x축 이름을 Day로 설정
plt.xlabel("Day")

# y축 이름을 Exercise Time (min)으로 설정
plt.ylabel("Exercise Time min")

# 그래프 출력하기
plt.show()
```
