#### line plot (선그래프)

```python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5, 6]
sales = [120, 150, 130, 170, 180, 210]

plt.plot(months, sales)
plt.show()
```

```python
# 선 색상 변경하기
plt.plot(months, sales, color="red")
plt.show()
```

```python
# 선 굵기 변경하기
# linewidth=3을 lw=3으로 줄여서 사용할 수도 있다.
plt.plot(months, sales, linewidth=3)
plt.show()
```

```python
# 선 스타일 변경하기
# - : 실선
# -- : 점선
# : 도트 점선
# -. : 일점 쇄선
plt.plot(months, sales, linestyle="--")
plt.show()
```

```python
# 마커 추가
# 마커 크기는 markersize로 조절
# o : 원
# s : 사각형
# ^ : 삼각형
# * : 별
# x : x로 표시
plt.plot(months, sales, marker="o", markersize=10)
plt.show()
```

```python
# 하나의 그래프에 여러 선 표현하기
months = [1, 2, 3, 4, 5, 6]

sales_2025 = [120, 150, 130, 170, 180, 210]
sales_2026 = [140, 160, 155, 190, 200, 230]

# label : 각 선의 이름을 지정
plt.plot(months, sales_2025, label="2025 sales")
plt.plot(months, sales_2026, label="2026 sales")

plt.legend() # 지정한 이름을 그래프에 표시
plt.show()
```

#### 실습문제 1

```python
# 월별 방문자 수를 선 그래프로 그리기
# 선 색상은 green
# 선 굵기는 2
# 마커는 원형(o), 크기는 8
# 그래프 제목은 Monthly Visitors
# x축 이름은 Month, y축 이름은 Visitors
# 그래프 출력하기

months = [1, 2, 3, 4, 5, 6]
visitors = [300, 450, 400, 550, 600, 750]

plt.plot(
    months, visitors,
    color="green", linewidth=2,
    marker="o", markersize=8
    )

plt.title("Monthly Visitors")
plt.xlabel("Month")
plt.ylabel("Visitors")

plt.show()
```

#### 실습문제 2

```python
# 2025년은 파란색 실선(-)
# 2026년은 빨간색 점선(--)
# 2025년 마커는 원형(o)
# 2026년 마커는 사각형(s)
# 두 선의 굵기는 2
# 각각 2025 Sales, 2026 Sales라는 label 지정
# plt.legend()로 범례 표시
# 제목은 Sales Comparison
# x축은 Month, y축은 Sales
# 그래프 출력

months = [1, 2, 3, 4, 5, 6]

sales_2025 = [100, 130, 120, 160, 180, 200]
sales_2026 = [120, 140, 150, 170, 210, 240]

plt.plot(months, sales_2025, color="blue",
    linestyle="-", marker="o", linewidth=2, label="2025 Sales")
plt.plot(months, sales_2026, color="red",
    linestyle="--", marker="s", linewidth=2, label="2026 Sales")

plt.title("Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()
plt.show()
```
