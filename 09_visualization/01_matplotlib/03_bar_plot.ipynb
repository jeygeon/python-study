#### 막대 그래프

```python
# 세로 막대 그래프
import matplotlib.pyplot as plt

departments = ["IT", "HR", "Sales", "Marketing"]
employees = [15, 8, 20, 12]

# plt.bar(x축, 높이)
plt.bar(departments, employees)

plt.title("Department Employees")
plt.xlabel("Department")
plt.ylabel("Employees")

plt.show()
```

```python
# 막대 색상 변경하기
plt.bar(departments, employees, color="skyblue")
plt.show()
```

```python
# 막대마다 색상을 다르게 적용할 수 있다.
plt.bar(departments, employees, color=["red", "blue", "green", "orange"])
plt.show()
```

```python
# width : 너비 변경
plt.bar(departments, employees, width=0.5)
plt.show()
```

```python
# plt.barh : 가로 막대 그래프
plt.barh(departments, employees, color="orange")

plt.title("Department Employees")
plt.xlabel("Employees")
plt.ylabel("Department")

plt.show()
```

```python
# 막대 위에 데이터 값 표현하기
departments = ["IT", "HR", "Sales", "Marketing"]
employees = [15, 8, 20, 12]

bars = plt.bar(departments, employees, color="skyblue")
plt.bar_label(bars)
plt.show()
```

---
#### 실습문제 1

```python
# plt.bar()를 이용해 세로 막대 그래프 그리기
# 막대 색상은 skyblue
# 막대 너비는 0.5
# 그래프 제목은 Department Employees
# x축은 Department, y축은 Employees
# 막대 위에 직원 수 표시하기
# 그래프 출력하기

departments = ["IT", "HR", "Sales", "Marketing"]
employees = [15, 8, 20, 12]

bars = plt.bar(departments, employees,
    color="skyblue", width=0.5
)

plt.title("Department Employees")
plt.xlabel("Department")
plt.ylabel("Employees")

plt.bar_label(bars)

plt.show()
```

---
#### 실습문제 2

```python
# plt.barh()를 이용해 가로 막대 그래프 그리기
# 막대 색상은 orange
# 그래프 제목은 Product Sales
# x축 이름은 Sales
# y축 이름은 Product
# 막대 끝에 판매량 표시하기
# 그래프 출력하기

products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
sales = [50, 120, 80, 40]

bars = plt.barh(products, sales, color="orange")

plt.title("Product Sales")
plt.xlabel("Sales")
plt.ylabel("Product")

plt.bar_label(bars)
plt.show()
```
