#### 💡 scatter_plot
산점도(scatter_plot)의 핵심 목적
- 두 변수간의 관계 파악
- 데이터 패턴 발견
- 변수간 상관관계 확인
- 데이터가 특정 방향으로 움직이는 추세 파악

<br/>

산점도에서는 아래 세가지 관계를 확인할 수 있다.
- **양의 상관관계** : X가 증가하면 Y도 증가하는 경향 (ex. 공부시간과 시험점수)
- **음의 상관관계** : X가 증가할 때 Y는 감소하는 경향 (ex. 운동량과 체지방률)
- **상관관계 없음** : 뚜렷한 관계가 보이지 않음 (ex. 신발 사이즈와 시험점수)

> 상관관계가 있다고 해서 무조건 인과관계가 있는것은 아니다.

```python
import matplotlib.pyplot as plt

# 데이터 생성
hours = [1, 2, 3, 4, 5, 6, 7]
scores = [45, 50, 60, 65, 75, 85, 95]

# 산점도 생성
plt.scatter(
    hours, scores,
    color="red",    # 점 색상
    s=100,          # 점 크기
    marker="o",     # 점 모양
    alpha=0.5       # 점 투명도
    )

plt.title("Study Hours vs Scores")
plt.xlabel("Study Hours")
plt.ylabel("Scores")

plt.show()
```

```python
# 여러 그룹 호출하기
import matplotlib.pyplot as plt

# IT 부서 데이터
it_hours = [2, 3, 4, 5, 6]
it_scores = [50, 55, 65, 70, 80]

# HR 부서 데이터
hr_hours = [1, 2, 3, 4, 5]
hr_scores = [45, 50, 55, 65, 70]

plt.scatter(it_hours, it_scores, color="blue", label="IT")
plt.scatter(hr_hours, hr_scores, color="red", label="HR")

plt.title("Department Study Hours vs Scores")
plt.xlabel("Study Hours")
plt.ylabel("Scores")

# 범례표시
plt.legend()

plt.show()
```

#### 실습문제

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6, 7]
scores = [45, 50, 60, 65, 75, 85, 95]

# plt.scatter()로 산점도 그리기
# 점 색상은 blue
# 점 크기는 100
# 점 모양은 o
# 그래프 제목은 Study Hours vs Scores
# X축 이름은 Study Hours
# Y축 이름은 Scores
# 그래프 출력하기

plt.scatter(
    hours, scores,
    color="blue", s=100, marker="o"
    )

plt.title("Study Hours vs Scores")
plt.xlabel("Study Hours")
plt.ylabel("Scores")

plt.show()
```

```python
# 남학생 데이터
male_hours = [1, 2, 3, 4, 5]
male_calories = [100, 150, 200, 250, 300]

# 여학생 데이터
female_hours = [1, 2, 3, 4, 5]
female_calories = [80, 120, 170, 210, 260]

# 남학생은 파란색(blue)으로 표시
# 여학생은 빨간색(red)으로 표시
# 두 그룹 모두 점 크기는 100
# 남학생과 여학생의 범례 표시
# 그래프 제목은 Exercise Hours vs Calories
# x축 이름은 Exercise Hours
# y축 이름은 Calories
# 그래프 출력

plt.scatter(male_hours, male_calories, s=100, color="blue", label="남학생")
plt.scatter(female_hours, female_calories, s=100, color="red", label="여학생")

# 한글 폰트 설정
plt.rcParams["font.family"] = "AppleGothic"
# 마이너스 기호 깨짐 방지
plt.rcParams["axes.unicode_minus"] = False

plt.title("Exercise Hours vs Calories")
plt.xlabel("Exercise Hours")
plt.ylabel("Calories")

plt.legend()
plt.show()
```
