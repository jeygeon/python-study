#### 💡 histogram
히스토그램이란 수치형 데이터가 특정 구간에 몇개씩 분포하는지 막대 형태로 나타내는 자료이다.

```python
# 히스토그램 그리기 : plt.hist()
import matplotlib.pyplot as plt

# Mac 한글 폰트 설정
plt.rcParams["font.family"] = "AppleGothic"
plt.rcParams["axes.unicode_minus"] = False

scores = [45, 52, 55, 60, 63, 65, 70, 75, 80, 90]

plt.hist(scores, bins=10, edgecolor="black", color="skyblue")

plt.title("Student Score Distribution")
plt.xlabel("Scores")
plt.ylabel("Frequency")

plt.show()
```

`bins=5`를 사용하면 자동으로 5개의 동일한 너비를 가진 구간을 만들어준다. 하지만 내가 원하는 기준으로 나누고 싶다면 경계 리스트를 `bins`에 전달하면 된다.

```python
scores = [
    45, 50, 55, 55, 60,
    62, 65, 68, 70, 72,
    75, 80, 85, 90, 95
]

plt.hist(scores, color="skyblue", edgecolor="black",
    bins=[40, 50, 60, 70, 80, 90, 100])

plt.title("Student Score Distribution")
plt.xlabel("Scores")
plt.ylabel("Frequency")

plt.show()
```

히스토그램에서 기본적으로 y가 Frequency. 즉 빈도수를 나타낸다. 하지만 `density` 옵션을 사용하면 데이터 갯수말고 확률 밀도(Probability Density)를 표시할 수 있다

```python
plt.hist(scores, color="orange", edgecolor="black", bins=5, density=True)

plt.title("Student Score Density")
plt.xlabel("Scores")
plt.ylabel("Density")

plt.show()
```

```python
scores = [
    45, 50, 55, 55, 60,
    62, 65, 68, 70, 72,
    75, 80, 85, 90, 95
]
```
