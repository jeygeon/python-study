#### 💡 pie chart (원형 그래프)
원형 그래프(pie chart)는 전체 데이터에서 각 항목이 차지하는 비율과 구성을 원 형태로 표현하는 그래프이다.

```python
# plt.pie() : 원형 그래프 그리기
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "AppleGothic"
plt.rcParams["axes.unicode_minus"] = False

departments = ["IT", "HR", "Sales", "Marketing"]
employees = [40, 20, 30, 10]

plt.pie(employees,
    labels=departments, # 구간별 라벨링ㄴ
    autopct="%1.1f%%", # 1.0 : 정수, 1.1 : 소수점 첫째짜리, 1.2 : 소수점 둘째자리
    colors=["skyblue", "orange", "lightgreen", "pink"], # 구가별 색깔 지정
    startangle=90, # 그래프가 시작하는 각도 지정,
    explode=[0.1, 0, 0, 0] # 특정 구간을 원에서 분리 (여기서는 0.1만큼)
    )

plt.title("Department Employee Distribution")

plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ["Electronics", "Clothing", "Food", "Books"]
sales = [40, 30, 20, 10]

wedges, texts, autotexts = plt.pie( # 범례추가
    sales,
    autopct="%1.1f%%",
    colors=["blue", "orange", "green", "red"],
    startangle=90,
    explode=[0.1, 0, 0, 0],
    shadow=True, # 그림자 추가
    pctdistance=0.7 # 퍼센트 위치 조정
)

# 범례 표시
plt.legend(
    wedges,
    categories,
    title="Categories"
)

plt.title("Category Sales Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "AppleGothic"
plt.rcParams["axes.unicode_minus"] = False

departments = ["IT", "HR", "Sales", "Marketing"]
employees = [40, 20, 30, 10]

plt.pie(
    employees,
    labels=departments,
    autopct="%1.1f%%",
    colors=["skyblue", "orange", "lightgreen", "pink"],
    startangle=90,
    explode=[0.1, 0, 0, 0]
)

plt.title("Department Employee Distribution")

plt.show()
```

```python
# 1. plt.pie()를 사용하기
# 2. 각 조각의 이름은 파이 차트에 직접 표시하지 않고 범례로 표시하기
# 3. 퍼센트는 소수점 첫째 자리까지 표시하기
# 4. 색상은 blue, orange, green, red
# 5. 시작 각도는 90
# 6. 그림자 활성화
# 7. 퍼센트 위치는 0.7
# 8. Python 조각만 0.1만큼 분리하기
# 9. 범례 제목은 Languages
# 10. 그래프 제목은 Programming Language Distribution

categories = ["Python", "Java", "JavaScript", "SQL"]
students = [40, 30, 20, 10]

wedges, text, autotext = plt.pie(students,
    autopct="%1.1f%%",
    colors  =["blue", "orange", "green", "red"],
    startangle=90,
    shadow=True,
    pctdistance=0.7,
    explode=[0.1, 0, 0, 0],
    )

plt.legend(
    wedges,
    categories,
    title="Languages"
)

plt.title("Programming Language Distribution")

plt.show()
```
