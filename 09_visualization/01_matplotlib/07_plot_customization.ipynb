#### Plot Customization : 그래프 꾸미기

```python
import matplotlib.pyplot as plt
```

```python
# 그래프 크기 조절저 : plt.figure()
# figsize=(4, 3) : 가로 4인치, 세로 3인치
plt.figure(figsize=(4, 3))

plt.plot([1, 2, 3, 4], [10, 20, 15, 30])

plt.title("Sales Trend")

plt.show()
```

```python
# 그래프 제목 및 글자 크기
plt.title("Monthly Sales", 
    fontsize=20,
    fontweight="bold",
    color="blue")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Sales", fontsize=12)
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 200]

plt.plot(months, sales, marker="o")

# xticks(rotation=45) : x축 눈금 글자를 45도 회전
# yticks([...]) : y축 눈금 글자를 직접 지정
plt.xticks(rotation=45)
plt.yticks([0, 50, 100, 150, 200, 250])

# 격자 표시
plt.grid(True,
    axis="y", # 세로 방향 격자
    linestyle="--", # 점선
    alpha=0.5) # 투명도

plt.show()
```

```python
# 그래프 범위 지정 : xlim(), ylim()
plt.plot(months, sales, marker="o")

# xlim() : x축의 범위
# ylim() : y축의 범위
plt.xlim(-0.5, 4.5)
plt.ylim(0, 250)

plt.show()
```

```python
# 데이터 위에 값 표시하기
plt.plot(months, sales, marker="o")

# i : x축 위치
# value + 5 : 데이터보다 조금 위쪽에 위치
# ha="center" : 가운데 정렬
for i, value in enumerate(sales):
    plt.text(i, value + 5, str(value), ha="center")

plt.show()
```
