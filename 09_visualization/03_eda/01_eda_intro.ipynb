### 💡 EDA (Exploratory Data Analysis)
`EDA`란 탐색적 데이터 분석이라는 뜻이다.

데이터를 본격적으로 분석하거나 머신러닝 모델을 만들기 전에 데이터를 여러 방향으로 살펴보면서 특징과 문제점을 파악하는 과정이다.

```python
import pandas as pd

data = {
    "name": ["Kim", "Lee", "Park", "Choi", "Jung"],
    "age": [25, 30, 28, 35, 30],
    "department": ["IT", "HR", "IT", "Sales", "HR"],
    "salary": [3000, 4000, 3500, 5000, 4000]
}

df = pd.DataFrame(data)

print(df)
```

위와 같은 데이터가 있다고 가정했을 때 아래와 같은 질문을 던질 수 있다.

- 직원은 총 몇명인지
- 어떤 부서가 있는지
- 나이가 너무 많은 직원이 있는지
- 급여가 비정상적으로 높은 직원이 있는지 등등..

이런 질문을 통해 데이터를 이해하는 과정이 `EDA`다.

#### EDA를 하는 주요 목적
- 데이터 구조 파악 : 행과 열, 컬럼명, 자료형
- 데이터 분포 확인 : 평균, 최솟값, 최대값, 분포
- 데이터 품질 확인 : 결측치, 중복값, 잘못된 데이터
- 이상치 탐색 : 일정범위를 벗어난 데이터
- 변수간 관계 파악 : 나이와 급여, 부서와 급여 등 관계
- 분석 방향 설정 : 어떤 분석 방법을 사용할지 결정

#### CRISP-DM
`CRISP-DM`은 데이터 분석 프로젝트를 진행하기 위한 대표적인 절차이다.

1. Business Understanding : 비즈니스 이해

    분석 목적과 해결해야 할 문제를 정의한다.

2. Data Understanding : 데이터 이해

    데이터를 수집하고 구조와 특징을 파악한다.

3. Data Preparation : 데이터 준비

    결측치, 중복값, 이상치 등을 처리하고 분석에 적합하게 가공한다.

4. Modeling : 모델링

    분석 목적에 맞는 통계적 방법이나 머신러닝 모델을 적용한다.

5. Evaluation : 평가

    분석 결과가 목적에 적합한지 평가한다.

6. Deployment : 배포

    분석 결과나 모델을 실제 업무 환경에 적용한다.
