```python
import pandas as pd

df = pd.read_csv("./datasets/titanic.csv")

# 데이터 앞부분 확인
df.head()
```

#### 데이터 구조 이해
1. 행(Row)

    하나의 데이터 단위를 의미. Titanic 데이터에서는 하나의 행

2. 열(Column)

    각 승객이 가진 개별 정보. survivied : 생존여부, pclass : 객실등급 등

3. 관측치(Observation)와 변수(Variable)

    관측치 : 분석 대상 하나에 대한 데이터 한 건
    변수 : 관측치마다 측정하거나 기록한 특성

#### 데이터 유형 구분
데이터는 크게 수치형과 범주형으로 구분하고, 각각을 다시 두 종류로 나누고 있다.
1. `수치형 데이터 (Numberical)` : 숫자로 크기나 양을 나타내는 데이터
    - **이산형** : 셀 수 있는 정수형 값 (ex. 판매량, 사람수, 형제자매 수)
    - **연속형** : 측정할 수 있고 소수점으로도 표현할 수 있는 값 (ex. 키, 몸무게, 온도, 운임)

2. `범주형 데이터 (Categorical)` : 특정 그룹이나 종류를 나타내는 데이터
    - **명목형** : 순사가 없는 데이터 (ex. 성별, 탑승항구, 혈액형)
    - **순서형** : 순서나 등급이 존재하는 데이터 (ex. 객실등급, 만족도, 학점 등급)

- **범주형 데이터** : sex(성별), embarked(탑승 항구), pclass(객실 등급), survived(생존, 사망 구분)
