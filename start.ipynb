## 많이쓰는 Jupyter notebook 단축키 모음

- `control + 엔터` : 코드 실행 (신규 코드 셀 생성 안함)
- `option + 엔터` : 코드 실행 (신규 코드 셀 생성)
- `command + a` : 해당 셀 전체 선택
- `command + c,v,z` : 복사, 붙여넣기, 실행 취소
- `command + /` : 해당 라인 주석처리 (다시 입력 시 주석 취소)

- `D + D` : 선택된 셀 삭제 (D 연속 두번 입력)
